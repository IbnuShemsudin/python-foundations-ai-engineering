```python
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Python:", sys.version)
print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)

numbers = np.array([10, 20, 30, 40, 50])

print("Average:", np.mean(numbers))
```

```python
plt.plot(numbers, marker="o")
plt.title("My First AI Engineering Data Plot")
plt.xlabel("Index")
plt.ylabel("Value")
plt.grid(True)
plt.show()
```
